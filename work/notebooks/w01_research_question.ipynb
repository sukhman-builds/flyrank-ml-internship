My provisional lane is **content-refresh review**.

I want to identify website pages whose search performance is declining and rank them for editorial review. This lane is useful because editors often have many pages but limited time to review them. A data-driven ranking can help them decide which pages deserve attention first.

I chose this lane because the starter dataset contains page-level information such as impressions, CTR, content age, update recency, and trend direction.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision is: **Which content pages should editors review first?**

The output will be a ranked list of pages that may need content review. An editor can inspect the highest-ranked pages and decide whether to update, improve, or leave them unchanged.

A wrong recommendation has two possible costs:
- A useful page may be ignored if it is ranked too low.
- An editor may waste time reviewing a page that does not need an update.

The model will support editorial decisions, not make automatic publishing decisions.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python

import os
import subprocess
import pandas as pd

REPO_DIR = "flyrank-ml-internship-starter"
DATA_PATH = os.path.join(
    REPO_DIR,
    "data",
    "raw",
    "content_refresh_anonymized.csv"
)

# Download starter repository if it is not already present
if not os.path.exists(DATA_PATH):
    subprocess.run(
        [
            "git", "clone", "--depth", "1",
            "https://github.com/flyrank-bih/flyrank-ml-internship-starter",
            REPO_DIR
        ],
        check=True
    )

# Load the anonymized dataset
df = pd.read_csv(DATA_PATH)

print("Total pages:", len(df))

# Count pages in each trend category
print("\nTrend distribution:")
print(df["trend_direction"].value_counts())

# Compare median impressions for declining and other pages
print("\nMedian impressions by trend:")
print(
    df.groupby("trend_direction")["impressions_90d"]
      .median()
      .round(2)
)

# Compare median CTR by trend
print("\nMedian CTR by trend:")
print(
    df.groupby("trend_direction")["ctr"]
      .median()
      .round(4)
)


```

### What I observed

1. The dataset contains 30,000 content pages, and 16,262 are labelled as `down`.
2. Pages labelled `down` have median impressions of 961, compared with 1,944.5 for `stable` pages.
3. Median CTR is 0.08 for `down` pages and 0.17 for `stable` pages.

These numbers suggest that declining pages form a substantial part of the dataset and show weaker median search performance than stable pages on these two measurements.

However, the `down` label describes an observed trend. It does not prove that a page needs an update or that updating it will improve performance.
