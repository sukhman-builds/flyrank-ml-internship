## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
My lane is content-refresh review.

The ML task is ranking: identify which content pages should be reviewed first based on their likelihood of currently declining.

The output will be a priority-ranked list of pages. This is more useful than only predicting yes/no because editors have limited time and need to decide which pages deserve attention first.

The main decision is: Which content page should an editor review first?
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
The target proxy is `is_declining_label`.

It is created from the observed `trend_direction` column:
- 1 = the page is marked as declining (`down`)
- 0 = the page has another trend direction.

This is an observed current-trend proxy, not a future decline label. The target is used for learning and evaluation only. `trend_direction` and `trend_pct` must never be used as model input features because they define or reveal the target.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
The primary success metric is Precision@50.

It measures how many of the 50 highest-priority pages are actually labelled as declining.

For example, Precision@50 = 0.70 means 35 out of the top 50 recommended pages are labelled as declining.

The final model should be evaluated using a client-holdout split so pages from the same client do not appear in both training and testing data.

A useful model should help editors find more relevant pages to review without wasting their limited review time.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import os
import sys
import subprocess
import pandas as pd
import numpy as np

REPO_DIR = "flyrank-ml-internship-starter"
DATA_PATH = os.path.join(
    REPO_DIR, "data", "raw", "content_refresh_anonymized.csv"
)

if not os.path.exists(DATA_PATH):
    subprocess.run(
        ["git", "clone", "--depth", "1",
         "https://github.com/flyrank-bih/flyrank-ml-internship-starter",
         REPO_DIR],
        check=True
    )

df = pd.read_csv(DATA_PATH)

df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

# Use pages with at least some impressions
lane_df = df[df["impressions_90d"] > 0].copy()

columns = [
    "content_id",
    "client_id",
    "content_type",
    "impressions_90d",
    "days_since_last_update",
    "content_age_days",
    "avg_position",
    "ctr",
    "trend_direction",
    "is_declining_label"
]

display(lane_df[columns].head(10))

print("Rows:", len(lane_df))
print("Columns:", len(columns))
print("One row = one anonymized content page")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
A fixed rule uses manually selected thresholds, such as marking pages for review when they are old and have high impressions.

However, content decline may depend on several interacting signals, including impressions, content age, update recency, CTR, and average search position.

A machine learning model can combine these signals and learn patterns that may be difficult to express through a few fixed conditions.

In the Week-1 pipeline, the Random Forest achieved Precision@50 of 0.740 compared with 0.240 for the baseline rule on the client-holdout evaluation.

This result suggests that ML can improve prioritization on this dataset, but it does not guarantee better performance in every setting. The model should only be preferred if its improvement holds on unseen client data.

The output supports editors in prioritizing pages for content review and possible refresh.
```
